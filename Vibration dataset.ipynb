```python
# =========================================================
# STEP 6: CREATE A VIBRATION FEATURE EXTRACTION FUNCTION
# =========================================================

import pandas as pd
import numpy as np


# ---------------------------------------------------------
# Function to extract statistical features from one
# vibration CSV file
# ---------------------------------------------------------

def extract_vibration_features(file_path):

    # Read the vibration data.
    # The first 10 lines contain metadata, so they are skipped.
    data = pd.read_csv(
        file_path,
        encoding="latin-1",
        skiprows=10
    )

    # Extract only the vibration acceleration values
    signal = data["Channel 3"].to_numpy()


    # -----------------------------------------------------
    # Calculate statistical vibration features
    # -----------------------------------------------------

    # Mean: average value of the vibration signal
    mean_value = np.mean(signal)

    # Standard deviation: amount of variation in vibration
    std_value = np.std(signal)

    # RMS (Root Mean Square):
    # represents the overall vibration energy/intensity
    rms_value = np.sqrt(np.mean(signal ** 2))

    # Peak:
    # largest absolute vibration amplitude
    peak_value = np.max(np.abs(signal))

    # Peak-to-Peak:
    # difference between maximum and minimum vibration
    peak_to_peak = np.max(signal) - np.min(signal)

    # Crest Factor:
    # ratio between peak vibration and RMS vibration
    # It can help identify sharp vibration peaks.
    if rms_value != 0:
        crest_factor = peak_value / rms_value
    else:
        crest_factor = 0


    # -----------------------------------------------------
    # Return all calculated features
    # -----------------------------------------------------

    return {
        "Mean": mean_value,
        "Std": std_value,
        "RMS": rms_value,
        "Peak": peak_value,
        "Peak_to_Peak": peak_to_peak,
        "Crest_Factor": crest_factor
    }


# ---------------------------------------------------------
# Test the function using our Ideal X-axis sample
# ---------------------------------------------------------

features = extract_vibration_features(sample_file)


print("Extracted vibration features:\n")

for feature_name, value in features.items():
    print(f"{feature_name}: {value:.6f}")
```

```python
# =========================================================
# STEP 8B: UPDATE THE FEATURE EXTRACTION FUNCTION
# =========================================================

import pandas as pd
import numpy as np


def extract_vibration_features(file_path):

    # -----------------------------------------------------
    # Read the vibration CSV file
    # -----------------------------------------------------

    # The first 10 lines contain sensor metadata.
    # Line 11 contains the column names.
    data = pd.read_csv(
        file_path,
        encoding="latin-1",
        skiprows=10
    )


    # -----------------------------------------------------
    # Automatically identify the vibration signal column
    # -----------------------------------------------------

    # The first column is "Sweeps".
    # The second column contains the vibration measurements.
    #
    # X-axis -> Channel 3
    # Y-axis -> Channel 2
    # Z-axis -> Channel 1
    signal_column = data.columns[1]

    # Extract the vibration values
    signal = data[signal_column].to_numpy(dtype=float)


    # -----------------------------------------------------
    # Calculate statistical vibration features
    # -----------------------------------------------------

    # Mean vibration value
    mean_value = np.mean(signal)

    # Standard deviation
    std_value = np.std(signal)

    # Root Mean Square (RMS)
    rms_value = np.sqrt(np.mean(signal ** 2))

    # Maximum absolute vibration amplitude
    peak_value = np.max(np.abs(signal))

    # Difference between maximum and minimum vibration
    peak_to_peak = np.max(signal) - np.min(signal)

    # Crest Factor = Peak / RMS
    if rms_value != 0:
        crest_factor = peak_value / rms_value
    else:
        crest_factor = 0


    # -----------------------------------------------------
    # Return the extracted features
    # -----------------------------------------------------

    return {
        "Mean": mean_value,
        "Std": std_value,
        "RMS": rms_value,
        "Peak": peak_value,
        "Peak_to_Peak": peak_to_peak,
        "Crest_Factor": crest_factor
    }


print("Feature extraction function updated successfully.")
```

```python
# =========================================================
# STEP 9: CHECK SAMPLE COUNTS FOR ALL MACHINE CONDITIONS
# =========================================================

import os


# List of machine conditions in the dataset
conditions = [
    "Ideal",
    "Offset_Pulley",
    "Cracking",
    "Wear"
]


# ---------------------------------------------------------
# Check X, Y and Z files for each condition
# ---------------------------------------------------------

for condition in conditions:

    print("\n----------------------------------------")
    print("Condition:", condition)
    print("----------------------------------------")

    # Construct paths for X, Y and Z folders
    x_folder = os.path.join(
        extract_path, "Data", condition, f"{condition}_X"
    )

    y_folder = os.path.join(
        extract_path, "Data", condition, f"{condition}_Y"
    )

    z_folder = os.path.join(
        extract_path, "Data", condition, f"{condition}_Z"
    )


    # Get only CSV files from each folder
    x_files = get_csv_files(x_folder)
    y_files = get_csv_files(y_folder)
    z_files = get_csv_files(z_folder)


    # Display number of files
    print("X-axis files:", len(x_files))
    print("Y-axis files:", len(y_files))
    print("Z-axis files:", len(z_files))


    # Check whether filenames match across all three axes
    files_match = x_files == y_files == z_files

    print("X, Y and Z filenames match:", files_match)


    # Display the filenames
    print("Samples:", x_files)
```

```python
# =========================================================
# STEP 10: BUILD THE COMPLETE FEATURE DATASET
# =========================================================

import os
import pandas as pd
import re


# ---------------------------------------------------------
# Function for natural sorting of sample filenames
# ---------------------------------------------------------

# This makes:
# M(1), M(2), M(3), ... M(10)
#
# instead of:
# M(1), M(10), M(2), ...

def sample_number(filename):

    # Extract the number inside M(...)
    match = re.search(r"M\((\d+)\)", filename)

    if match:
        return int(match.group(1))

    return 0


# ---------------------------------------------------------
# List to store every feature row
# ---------------------------------------------------------

all_samples = []


# ---------------------------------------------------------
# Process each machine condition
# ---------------------------------------------------------

for condition in conditions:

    print(f"Processing: {condition}")

    # Create paths for X, Y and Z folders
    x_folder = os.path.join(
        extract_path, "Data", condition, f"{condition}_X"
    )

    y_folder = os.path.join(
        extract_path, "Data", condition, f"{condition}_Y"
    )

    z_folder = os.path.join(
        extract_path, "Data", condition, f"{condition}_Z"
    )


    # -----------------------------------------------------
    # Get the CSV filenames
    # -----------------------------------------------------

    x_files = sorted(
        get_csv_files(x_folder),
        key=sample_number
    )

    y_files = sorted(
        get_csv_files(y_folder),
        key=sample_number
    )

    z_files = sorted(
        get_csv_files(z_folder),
        key=sample_number
    )


    # Safety check:
    # Make sure corresponding X, Y and Z filenames match
    assert x_files == y_files == z_files, \
        f"X/Y/Z file mismatch found in {condition}"


    # -----------------------------------------------------
    # Process each measurement
    # -----------------------------------------------------

    for filename in x_files:

        # Complete file paths
        x_file = os.path.join(x_folder, filename)
        y_file = os.path.join(y_folder, filename)
        z_file = os.path.join(z_folder, filename)


        # Extract features from each vibration axis
        x_features = extract_vibration_features(x_file)
        y_features = extract_vibration_features(y_file)
        z_features = extract_vibration_features(z_file)


        # Create one row for this measurement
        row = {}


        # Add X-axis features
        for feature_name, value in x_features.items():
            row["X_" + feature_name] = value


        # Add Y-axis features
        for feature_name, value in y_features.items():
            row["Y_" + feature_name] = value


        # Add Z-axis features
        for feature_name, value in z_features.items():
            row["Z_" + feature_name] = value


        # Add sample information
        row["Sample"] = filename.replace(".csv", "")
        row["Condition"] = condition


        # Store the completed row
        all_samples.append(row)


# ---------------------------------------------------------
# Convert all rows into a pandas DataFrame
# ---------------------------------------------------------

feature_dataset = pd.DataFrame(all_samples)


# ---------------------------------------------------------
# Display results
# ---------------------------------------------------------

print("\nFeature dataset created successfully!")

print("\nDataset shape:")
print(feature_dataset.shape)

print("\nNumber of samples in each condition:")
print(feature_dataset["Condition"].value_counts())

print("\nFirst 5 rows:")
display(feature_dataset.head())
```

```python
# =========================================================
# STEP 11: COMPARE FEATURES ACROSS MACHINE CONDITIONS
# =========================================================

# ---------------------------------------------------------
# Select only the numerical vibration features
# ---------------------------------------------------------

# "Sample" is only an identifier.
# "Condition" is our target label.
# Therefore, they are excluded from the feature list.
feature_columns = [
    column
    for column in feature_dataset.columns
    if column not in ["Sample", "Condition"]
]


# ---------------------------------------------------------
# Calculate the average value of every feature
# for each machine condition
# ---------------------------------------------------------

condition_means = (
    feature_dataset
    .groupby("Condition")[feature_columns]
    .mean()
)


# Display the averages
print("Average feature values for each machine condition:\n")

display(condition_means.round(3))


# ---------------------------------------------------------
# Check for missing values in the complete dataset
# ---------------------------------------------------------

print("\nTotal missing values in each column:\n")

print(feature_dataset.isnull().sum())


# ---------------------------------------------------------
# Check for infinite numerical values
# ---------------------------------------------------------

import numpy as np

infinite_values = np.isinf(
    feature_dataset[feature_columns]
).sum()

print("\nInfinite values in numerical features:\n")

print(infinite_values)
```

```python
# =========================================================
# STEP 12: PREPARE X AND y AND SPLIT THE DATASET
# =========================================================

from sklearn.model_selection import train_test_split


# ---------------------------------------------------------
# Create the input features (X)
# ---------------------------------------------------------

# Use only the 18 numerical vibration features.
# "Sample" is just an identifier and should NOT be used
# to train the machine learning model.
X = feature_dataset[feature_columns]


# ---------------------------------------------------------
# Create the target variable (y)
# ---------------------------------------------------------

# The machine condition is what the Random Forest
# will learn to predict.
y = feature_dataset["Condition"]


# ---------------------------------------------------------
# Split the dataset into training and testing sets
# ---------------------------------------------------------

# 80% of the samples are used for training.
# 20% are kept aside for testing.
#
# stratify=y keeps approximately the same class
# distribution in both training and testing sets.
#
# random_state=42 makes the split reproducible.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ---------------------------------------------------------
# Display the results
# ---------------------------------------------------------

print("Total samples:", len(X))

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


print("\nTraining class distribution:")
print(y_train.value_counts())


print("\nTesting class distribution:")
print(y_test.value_counts())


print("\nNumber of input features:", X_train.shape[1])
```

```python
# =========================================================
# STEP 13: RANDOM FOREST HYPERPARAMETER TUNING
# =========================================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold


# ---------------------------------------------------------
# 1. Create the base Random Forest model
# ---------------------------------------------------------

rf_base = RandomForestClassifier(
    random_state=42,
    class_weight="balanced"
)


# ---------------------------------------------------------
# 2. Define the hyperparameters to test
# ---------------------------------------------------------

param_grid = {

    # Number of trees in the forest
    "n_estimators": [100, 200, 300, 500],

    # Maximum depth of each tree
    "max_depth": [None, 3, 5, 7, 10, 15],

    # Minimum samples required to split a node
    "min_samples_split": [2, 3, 4, 5, 6],

    # Minimum samples required in a leaf node
    "min_samples_leaf": [1, 2, 3, 4],

    # Number of features considered for each split
    "max_features": ["sqrt", "log2", None],

    # Method used to measure split quality
    "criterion": ["gini", "entropy"],

    # Whether bootstrap sampling is used
    "bootstrap": [True, False]
}


# ---------------------------------------------------------
# 3. Create 5-fold stratified cross-validation
# ---------------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ---------------------------------------------------------
# 4. Create RandomizedSearchCV
# ---------------------------------------------------------

random_search = RandomizedSearchCV(
    estimator=rf_base,
    param_distributions=param_grid,
    n_iter=30,
    scoring="f1_macro",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)


# ---------------------------------------------------------
# 5. Search for the best Random Forest settings
# ---------------------------------------------------------

random_search.fit(X_train, y_train)


# ---------------------------------------------------------
# 6. Display the best parameters
# ---------------------------------------------------------

print("\nBest Random Forest Parameters:\n")

for parameter, value in random_search.best_params_.items():
    print(f"{parameter}: {value}")


# ---------------------------------------------------------
# 7. Display the best cross-validation F1 score
# ---------------------------------------------------------

print(
    "\nBest Cross-Validation F1 Score:",
    f"{random_search.best_score_ * 100:.2f}%"
)


# ---------------------------------------------------------
# 8. Store the best model
# ---------------------------------------------------------

rf_model = random_search.best_estimator_

print("\nBest Random Forest model selected successfully!")
```

```python
# =========================================================
# STEP 14: EVALUATE THE TUNED RANDOM FOREST MODEL
# =========================================================

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Make predictions using the tuned Random Forest
y_pred = rf_model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Tuned Random Forest Test Accuracy:")
print(f"{accuracy * 100:.2f}%")

# Classification report
print("\nClassification Report:\n")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))

# Confusion matrix
print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))
```

```python
# =========================================================
# STEP 15: 5-FOLD CROSS-VALIDATION OF TUNED RANDOM FOREST
# =========================================================

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier
import numpy as np


# ---------------------------------------------------------
# 1. Create the tuned Random Forest model
# ---------------------------------------------------------

tuned_rf_model = RandomForestClassifier(
    n_estimators=500,
    min_samples_split=4,
    min_samples_leaf=4,
    max_features="log2",
    max_depth=7,
    criterion="entropy",
    bootstrap=False,
    class_weight="balanced",
    random_state=42
)


# ---------------------------------------------------------
# 2. Create 5-fold stratified cross-validation
# ---------------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ---------------------------------------------------------
# 3. Define evaluation metrics
# ---------------------------------------------------------

scoring = {
    "accuracy": "accuracy",
    "precision": "precision_macro",
    "recall": "recall_macro",
    "f1": "f1_macro"
}


# ---------------------------------------------------------
# 4. Perform 5-fold cross-validation
# ---------------------------------------------------------

cv_results = cross_validate(
    tuned_rf_model,
    X,
    y,
    cv=cv,
    scoring=scoring,
    n_jobs=-1
)


# ---------------------------------------------------------
# 5. Display results for each fold
# ---------------------------------------------------------

print("5-Fold Cross-Validation Results:\n")

for i in range(5):

    print(f"Fold {i + 1}:")

    print(
        f"Accuracy  : "
        f"{cv_results['test_accuracy'][i] * 100:.2f}%"
    )

    print(
        f"Precision : "
        f"{cv_results['test_precision'][i] * 100:.2f}%"
    )

    print(
        f"Recall    : "
        f"{cv_results['test_recall'][i] * 100:.2f}%"
    )

    print(
        f"F1 Score  : "
        f"{cv_results['test_f1'][i] * 100:.2f}%"
    )

    print()


# ---------------------------------------------------------
# 6. Calculate average performance
# ---------------------------------------------------------

print("Average Cross-Validation Performance:\n")

print(
    f"Average Accuracy  : "
    f"{np.mean(cv_results['test_accuracy']) * 100:.2f}%"
)

print(
    f"Average Precision : "
    f"{np.mean(cv_results['test_precision']) * 100:.2f}%"
)

print(
    f"Average Recall    : "
    f"{np.mean(cv_results['test_recall']) * 100:.2f}%"
)

print(
    f"Average F1 Score  : "
    f"{np.mean(cv_results['test_f1']) * 100:.2f}%"
)

print(
    f"Accuracy Standard Deviation: "
    f"{np.std(cv_results['test_accuracy']) * 100:.2f}%"
)
```
